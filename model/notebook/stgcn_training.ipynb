```python
%%writefile /content/debug_hang3.py
import os, time
os.environ.setdefault('RANK', '0')
os.environ.setdefault('WORLD_SIZE', '1')
os.environ.setdefault('LOCAL_RANK', '0')
os.environ.setdefault('MASTER_ADDR', '127.0.0.1')
os.environ.setdefault('MASTER_PORT', '29500')

print("start", flush=True)

from mmcv import Config
cfg = Config.fromfile('/content/pyskl/configs/stgcn/stgcn_lite_hmdb51_hrnet/j.py')

import torch
from mmcv.runner import init_dist
init_dist('pytorch', **cfg.dist_params)
print("init_dist done", flush=True)

from pyskl.datasets import build_dataset, build_dataloader
dataset = build_dataset(cfg.data.train)
loader = build_dataloader(
    dataset,
    videos_per_gpu=cfg.data.videos_per_gpu,
    workers_per_gpu=cfg.data.workers_per_gpu,
    seed=cfg.get('seed', None))
batch = next(iter(loader))
print("batch fetched", flush=True)

from pyskl.models import build_model
model = build_model(cfg.model).cuda()
print("model on GPU", flush=True)

t0 = time.time()
from torch.nn.parallel import DistributedDataParallel
model = DistributedDataParallel(model, device_ids=[0])
print("DDP wrap done", flush=True)

kp = batch['keypoint'].cuda()
lbl = batch['label'].cuda()
print("moved batch to GPU", flush=True)

t0 = time.time()
losses = model(kp, lbl, return_loss=True)
print("forward pass done", flush=True)
print(losses, flush=True)

loss = sum(v for v in losses.values() if isinstance(v, torch.Tensor) and v.requires_grad)
loss.backward()
print("backward pass done", flush=True)
```

```python
%%writefile /content/debug_hang4.py
import os, time
os.environ.setdefault('RANK', '0')
os.environ.setdefault('WORLD_SIZE', '1')
os.environ.setdefault('LOCAL_RANK', '0')
os.environ.setdefault('MASTER_ADDR', '127.0.0.1')
os.environ.setdefault('MASTER_PORT', '29500')

from mmcv import Config
cfg = Config.fromfile('/content/pyskl/configs/stgcn/stgcn_lite_hmdb51_hrnet/j.py')

from mmcv.runner import init_dist
init_dist('pytorch', **cfg.dist_params)

from pyskl.datasets import build_dataset, build_dataloader
dataset = build_dataset(cfg.data.train)
loader = build_dataloader(
    dataset,
    videos_per_gpu=cfg.data.videos_per_gpu,
    workers_per_gpu=cfg.data.workers_per_gpu,
    seed=cfg.get('seed', None))

print(f"loader ready, {len(loader)} batches total", flush=True)

t0 = time.time()
for i, batch in enumerate(loader):
    dt = time.time() - t0
    print(f"batch {i}: {dt:.2f}s", flush=True)
    t0 = time.time()
    if i >= 120:
        break
```
